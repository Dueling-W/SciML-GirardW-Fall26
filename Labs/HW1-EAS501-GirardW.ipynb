```python
import numpy as np
from scipy.linalg import lu
import pandas as pd
```

```python
# construct matrix A using loops
def construct_A(A): 

    # 1 to 10 (excludes 11)
    for r in range(10): 
        for c in range(10): 
            # shift to 1-based indexing
            i = r +1 
            j = c + 1

            if(i == j and (i>=2 and j <= 9)):
                A[r, c] = -5/2 
            elif(i==j==1 or i==j==10):
                A[r, c] = -29/12 
            elif(i-j==1 or j-i ==1):
                A[r, c] = 4/3 
            elif(i-j==2 or j-i ==2):
                A[r, c] = -1/12
            else:
                A[r, c] = 0

    return A


# create empty matrix A of size 10x10
A = np.zeros(shape=(10, 10))
A = construct_A(A)

A_df = pd.DataFrame(A, index=range(1, 11), columns=range(1, 11))
A_df
```

```python
import pandas as pd

# compute LU decomposition
L, U = lu(A, permute_l=True)

# convert to dataframe for printing
# set index/columns labels to 1-10
L = pd.DataFrame(L, index=range(1, 11), columns=range(1, 11))
print('L Matrix: \n')
L
```

```python
U = pd.DataFrame(U, index=range(1, 11), columns=range(1,11))
print('U Matrix: \n')
U
```

The matricies L and U have been printed above, with proper index labeling from 1-10. In total, there are 100 elements. Note that U has the lower triangle and L has the upper triangle form, with 1's along the diagonal for L (known as Doolittle's algorithm/method). Now, I just need to get the determinant from U.

```python
import math
# get determinant for A from U by multiplying its diagonal entries
# note that U is the result of Gaussian elimination

# extract diagonals and multiply
diag_U = np.diagonal(U)
det_A = math.prod(diag_U)

print(f'Determinant of A obtained through LU decomposition and Gaussian elimination is: {det_A:.3f}') # round slightly
```

```python
# double check result with numpy's built in det method 
# this is a sanity check to make sure my code worked

print(f'Determinant of A through numpys method is {np.linalg.det(A):.3f}')
```

Great, they match exactly. Showing that LU decomposition --> multiplying U's diagonals is a valid (and simple) way of finding the determinant of a matrix.
