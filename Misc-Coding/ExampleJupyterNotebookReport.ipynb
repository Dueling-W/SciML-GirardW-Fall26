# Problem 2: analysis of a data file  (**50 points**)

Scott's CPU monitoring script (task 1 of lab 2) produced a data file called CPUData.dat

In this section, we will see how Python code can be easily included in the notebook for quick data analysis. Unlike the above example, we will not execute the python code. This works because the following cells are "code cells", whereas the above ones were markdown.

````python
import numpy as np
from matplotlib import pyplot as plt

data = np.loadtxt('CPUData.dat') # <- Will automattically ignore lines starting with #

x = data[:,0]
y = data[:,1]

plt.figure()
plt.plot(x,y,'r--',label='Google Chrome')
plt.xlabel('Unix Coordinate Time (seconds)')
plt.ylabel('CPU Usage (percentage)')

# NOTE: Notice we are putting a latex equation into the title
# by using "raw strings" (notice the small "r" in front of the string)
plt.title(r'CPU Usage (fancy latex $\int_{1}^{\infty} \frac{1}{x} dx$)')

plt.legend(loc=3)
plt.show()
# The next line is commented out. Uncomment it if you'd like to save the image as a png
#plt.savefig("cpu_vs_time.png")
````

In this next example, we need to use Jupyter notebook "magic" command "%%bash". This will run the block of commands after it  inside of your shell. 

This is useful for more complicated tasks. 

Here we create a simple C program using echo. you should have a new file called "hello.c" create in this directory.

````python
%%bash
echo '#include <stdio.h>
int main() {
    printf("Hello, World!\\n");
    return 0;
}' > hello.c
````

Now we can use the Unix command Cat to show the contents of the new hello.c file directly inside of the notebook.

This is another convenient way to include code into your Jupyter notebook.

````python
cat hello.c
````
