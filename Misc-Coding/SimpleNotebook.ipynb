\begin{align*}
  \int_{-1}^1\frac{1}{1+x^2}dx \\
\end{align*}

```python
import scipy.integrate as integrate
# Define the function
def f(x):    
    return 1 / (1 + x**2)
    
# Compute the definite integral from -1 to 1
result, error = integrate.quad(f, -1, 1)

# Print the result
print(f"Integral result: {result}")
print(f"Estimated error: {error}")
```
